```python
import neurokit2 as nk
import pandas as pd
import matplotlib as plt
import seaborn as sns

#
sns.set(style = "whitegrid")
ecg50 = nk.ecg_simulate(duration = 10, noise = 0.05, heart_rate = 50)
ecg100 = nk.ecg_simulate(duration = 10, noise = 0.01, heart_rate = 100)
ecg150 = nk.ecg_simulate(duration = 10, noise = 0.1, heart_rate = 150)
ecg_df = pd.DataFrame({"ECG_50": ecg50, "ECG_100": ecg100, "ECG_150": ecg150 })

nk.signal_plot(ecg_df, subplots = True)
```

```python
emg2 = nk.emg_simulate(duration = 10, burst_number = 2, burst_duration = 1.0)
emg4 = nk.emg_simulate(duration = 10, burst_number = 4, burst_duration = 1.5)
emg6_long = nk.emg_simulate(duration = 10, burst_number = 6, burst_duration = 0.5)
# Visualizar

emg_df = pd.DataFrame({"EMG_2": emg2, "EMG_4": emg4, "EMG_6": emg6_long})
nk.signal_plot(emg_df, subplots = True)
```
